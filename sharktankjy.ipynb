```python
import pandas as pd

# Load dataset using full path
df = pd.read_csv(r"C:\Users\nalab\Downloads\SharkTankIndiaDataset.csv")

# Check first few rows
df.head()
```

```python
# Show all column names
print(df.columns.tolist())
```

```python
# Summary statistics for numeric columns
df.describe()

# Count missing values per column
df.isnull().sum()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
print("Duplicates removed successfully!")
```

```python
# Clean column names: remove spaces, make lowercase, replace spaces with underscores
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
print("Column names standardized!")
```

```python
print(df.info())
print(df.head())
```

```python
# Save cleaned dataset to a new CSV file
df.to_csv(r"C:\Users\nalab\Downloads\sharktank_clean.csv", index=False)
print("Cleaned dataset saved successfully at C:\\Users\\nalab\\Downloads\\sharktank_clean.csv")
```
